```python
# Get student information
name = input("Enter student name: ").strip()
course = input("Enter course name: ").strip()

# Open the file in append mode
with open("student_list.txt", "a") as file:

    # Write the student information to the file
    file.write(name + " - " + course + "\n")

# Confirm that the student was saved
print("Student saved successfully!")
```

```python
#Updating student information
# Ask which student to update
student_name = input("Enter student name to update: ").strip()

# Ask for the new name
new_name = input("Enter new student name: ").strip()

# Create a temporary list for the updated records
students = []

# Keep track of whether the student was found
found = False

# Read the existing file
with open("student_list.txt", "r") as file:

    # Check each student
    for line in file:

        # Remove whitespace and compare the student name
        if line.strip() == student_name:

            # Replace the old name with the new name
            students.append(new_name)
            found = True

        else:

            # Keep the existing student
            students.append(line.strip())

# Rewrite the file with the updated records
with open("student_list.txt", "w") as file:

    # Write each student back to the file
    for student in students:
        file.write(student + "\n")

# Display the result
if found:
    print("Student found and updated successfully!")
else:
    print("Student not found!")
```

```python
# Ask which student to delete
student_name = input("Enter student name to delete: ").strip()

# Create a temporary list for the remaining students
students = []

# Keep track of whether the student was found
found = False

# Read the existing file
with open("student_list.txt", "r") as file:

    # Check each student
    for line in file:
        student = line.strip()

        # Skip the student we want to delete
        if student == student_name:
            found = True
        else:
            # Keep all other students
            students.append(student)

# Rewrite the file with the remaining students
with open("student_list.txt", "w") as file:

    # Write each remaining student back to the file
    for student in students:
        file.write(student + "\n")

# Display the result
if found:
    print("Student deleted successfully!")
else:
    print("Student not found!")
```
