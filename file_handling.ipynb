```python
# Create and write student names to the file
# Read the student names from the file

file = open("student.txt", "r")

content = file.read()

print(content)

file.close()
```

```python
#better Approach using "with"

with open("student.txt", "r") as file:
    content = file.read()

print(content)
```

```python
# Create the file and write student names
with open("student_list.txt", "w") as file:
    file.write("John\n")
    file.write("Mary\n")
    file.write("Peter\n")

# Now read the file
with open("student_list.txt", "r") as file:
    content = file.read()

# Display the contents
print(content)
```

```python
# Open the student file


with open("student_list.txt", "r") as file:
    #Loop through each student
    for line in file:
        print("Student: ", line.strip())




```

```python
# Add new students
with open("student_list.txt", "a") as file:
    file.write("Cosmas\n")
    file.write("Sarah\n")
    file.write("David\n")

# Read all students
with open("student_list.txt", "r") as file1:
    for line in file1:
        print("Student: ", line.strip())
```

```python
#counting students
student_count = 0
with open("student_list.txt", "r") as file:
    for line in file:
        if line.strip():
            student_count += 1

# Display the total number of students
print("Total students:", student_count)
```

```python
# Create an empty list for students
students = []

# Open the student file

with open ("student_list.txt", "r") as file:
    for line in file:
        if line.strip():
            students.append(line.strip())

print(students)
```
