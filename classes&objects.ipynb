```python
#using classes to build multiple choice questions

from Question import Question

question_prompts = [
    "What is the capital of Kenya?\n(a) Nairobi\n(b) Mombasa\n(c) Kisumu\n(d) Nakuru\n",
    "Which language is used to create Python programs?\n(a) Java\n(b) Python\n(c) C++\n(d) HTML\n",
    "What is 10 + 5?\n(a) 10\n(b) 15\n(c) 20\n(d) 25\n",
    "Which one is a Python data type?\n(a) List\n(b) Browser\n(c) Website\n(d) Server\n",
    "What does CPU stand for?\n(a) Central Processing Unit\n(b) Computer Personal Unit\n(c) Central Program Utility\n(d) Computer Processing User\n"
]

questions = [
    Question(question_prompts[0], "a"),
    Question(question_prompts[1], "b"),
    Question(question_prompts[2], "c"),
    Question(question_prompts[3], "a"),
    Question(question_prompts[4], "a")
]

def run_question_test(questions):
    score = 0
    for question in questions:
        answer = input(question.prompt)
        if answer == question.answer:
            score = score + 1
    print("You got " + str(score) + " correct questions out of " + str(len(questions)))

run_question_test(questions)
```

```python
#Inheritance
#I have the chef file
from chef import Chef

class ItalianChef(Chef):
    def make_pasta(self):
        print(f"{self.name} is making authentic Italian pasta.")
```

```python
class PastryChef(Chef):
    def bake_cake(self):
        print(f"{self.name} is baking a delicious cake.")
```

```python
class SushiChef(Chef):
    def prepare_sushi(self):
        print(f"{self.name} is preparing fresh sushi.")
```

```python
chef1 = ItalianChef("John", 10)
chef2 = PastryChef("Mary", 7)
chef3 = SushiChef("Peter", 12)


chef1.cook()
chef2.cook()
chef3.clean_kitchen()

chef1.make_pasta()
chef2.bake_cake()
chef3.prepare_sushi()
```
