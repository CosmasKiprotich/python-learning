```python
#Using comparisons in if statements
#determining the largest number and smallest number


def maximum_num(num1, num2, num3):
    if num1 >= num2 and num1 >= num3:
        return num1
    elif num2 >= num1 and num2 >= num3:
        return num2
    else:
        return num3

def minimum_num(num1, num2, num3):
    if num1 <= num2 and num1 <= num3:
        return num1
    elif num2 <= num1 and num2 <= num3:
        return num2
    else:
        return num3


print("The maximum number is " + str(maximum_num(2,2,4)))
print("The minimum number is " + str(minimum_num(2,2,4)))

print("The maximum number is " + str(maximum_num(300,20000,4400)))
print("The minimum number is " + str(minimum_num(300,20000,4400)))
```

```python
#building a basic calculator
#user inputs

num1 = float(input("Enter your first number: "))
op = input("Enter your operation: ")
num2 = float(input("Enter your second number: "))

if op == "+":
    print(num1 + num2)
elif op == "-":
    print(num1 - num2)
elif op == "*":
    print(num1 * num2)
elif op == "/":
    print(num1 / num2)
else:
    print("Invalid operation")
```
