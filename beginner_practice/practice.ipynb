```python
#Using comparisons in if statements
#determining the largest number and smallest number


def maximum_num(numb1, numb2, num3):
    if numb1 >= numb2 and numb1 >= num3:
        return num1
    elif numb2 >= num1 and numb2 >= num3:
        return numb2
    else:
        return num3

def minimum_num(numb1, numb2, numb3):
    if num1 <= numb2 and numb1 <= numb3:
        return numb1
    elif num2 <= numb1 and numb2 <= numb3:
        return numb2
    else:
        return numb3


print("The maximum number is " + str(maximum_num(2,2,4)))
print("The minimum number is " + str(minimum_num(2,2,4)))

print("The maximum number is " + str(maximum_num(300,20000,4400)))
print("The minimum number is " + str(minimum_num(300,20000,4400)))
```

```python
#building a basic calculator
#user inputs

num1 = float(input("Enter your first number: "))
op = input("Enter your operation: ")
num2 = float(input("Enter your second number: "))

if op == "+":
    print(num1 + num2)
elif op == "-":
    print(num1 - num2)
elif op == "*":
    print(num1 * num2)
elif op == "/":
    print(num1 / num2)
else:
    print("Invalid operation")
```

```python
#Buidling a guessing game using while loops

secret_name = "cosmas"

guess = ""

guess_count = 0

guess_limit = 3

out_of_guesses = False

while guess != secret_name and not out_of_guesses:
    if guess_count < guess_limit:
        guess = input("Enter your guess: ")
        guess_count = guess_count + 1
    else:
        out_of_guesses = True


if out_of_guesses:
    print("You are out of guesses, YOU LOSE")
else:
    print("Correct you won the game!")
```

```python
#using for loop to do power computation

def raise_to_power(base_number, exponent):
    result = 1
    for index in range(exponent):
        result = result * base_number
    return result


raise_to_power(10, 3)
```
