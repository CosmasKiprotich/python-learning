```python
# Challenge 6 — Add a Menu
# Challenge 7 — Connect the CRUD Operations
# Challenge 8 — Improve the View Students Option
# Challenge 9 — Validate Student Input
# Challenge 10 — Prevent Duplicate Student IDs
# Challenge 11 — Handle Empty Student List
# Challenge 12 — Improve Search Student
# Challenge 13 — Validate Grade
# Challenge 14 — Improve Update Grade
# Challenge 15 — Confirm Student Deletion
# Challenge 16 — Pause Before Returning to Menu
# Challenge 17 — Validate Course Input
# Challenge 18 — Validate Student Name
# Challenge 19 — Validate Grade When Adding a Student
# Challenge 20 — Clean Up User Input
# Challenge 21 — Validate Student ID
# Challenge 22 — Clean Up Menu Choice
# Challenge 23 — Clean Up Search Student ID
#Challenge 26 — Refactor the Student Management System using functions
#Creating a Student Management System menu.
# ===== Student Management System =====
# 1. View Students
# 2. Search Student
# 3. Add Student
# 4. Update Grade
# 5. Delete Student
# 6. Exit

print("_________Student Management System Menu_________")


#functions
def view_students():
    print("View all students")
    print("========== Student List ==========")
    if not students:
        print("No students found.")
    else:
        for student in students:
            print("ID: ", student["student_id"])
            print("Name: ", student["name"])
            print("Age: ", student["age"])
            print("Course: ", student["course"])
            print("Student Grade: ", student["grade"])
            print("----------------------------------")
            print("                                  ")
            print("                                  ")

def search_student():
    print("Search for students")
    while True:
        std_id = input("Enter student id of the student you want to check: ").strip()
        if not std_id:
            print("Student ID cannot be empty")
            continue
        break
    found_student = False
    if std_id:
        for student in students:
            if student["student_id"] == std_id:
                found_student = True
                print("Student Found")
                print("__________Student Details_____________")
                print("ID: ", student["student_id"])
                print("Name: ", student["name"])
                print("Age: ", student["age"])
                print("Course: ", student["course"])
                print("Grade: ", student["grade"])
                break
    if not found_student:
        print("Student Not Found")

def add_student():
        print("Add student to database")

        while True:
            student_id = input("Enter student id: ").strip()
            found_student = False
            if not student_id:
                print("Student ID cannot be empty")
                continue

            for student in students:
                if student["student_id"] == student_id:
                    found_student = True
                    print("Student ID already exists")
                    break
            if found_student:
                continue
            break

        while True:
            name = input("Enter student name: ").strip()
            if name:
                break
            else:
                print("Student name cannot be empty")

        while True:
            try:
                age = int(input("Enter student age: "))

                if age > 0 and age <= 100:
                    break

            except ValueError:
                print("Invalid Age")

        while True:
            course = input("Enter student course: ").strip()
            if course:
                break
            else:
                print("Course cannot be empty")

        while True:
            grade = input("Enter student grade: ")
            grade = grade.upper()
            if grade in ["A","B","C","D","F"]:
                break
            else:
                print("Grade must be A, B, C, D or F")
                print("Invalid Grade")

        new_student = {
            "student_id": student_id,
            "name": name,
            "age": age,
            "course": course,
            "grade": grade
        }
        students.append(new_student)

        print("Student Added Successfully")
        print("_________New Student Details_____________")
        print("Name: ", new_student["name"])

def update_grade():
    print("Update a student's grade")
    std_id = input("Enter student id of the student you want to change grade: ")
    found_student = False

    for student in students:
        if student["student_id"] == std_id:
            found_student = True
            print("Student Found")

            print("Current Grade: ", student["grade"])

            while True:
                grade = input("Enter new student grade: ")
                grade = grade.upper()
                if grade in ["A", "B", "C", "D", "F"]:
                    break
                else:
                    print("Invalid Grade")

            student["grade"] = grade
            print("New Grade: ", student["grade"])

            print("Grade Changed Successfully")
            break
    if not found_student:
        print("Student Not Found")

def delete_student():
    print("Delete a student")
    std_id = input("Enter student id of the student you want to delete: ")
    found_student = False

    for student in students:
        if student["student_id"] == std_id:
            found_student = True
            print("__________Student Details_____________")
            print("ID: ", student["student_id"])
            print("Name: ", student["name"])
            print("Course: ", student["course"])
            print("Grade: ", student["grade"])
            confirmation = input("Are you sure you want to delete this student? (yes/no): ")
            confirmation = confirmation.lower()
            if confirmation == "yes":
                students.remove(student)
                print("Student Deleted Successfully")
                break
            else:
                print("Deletion Cancelled")
                break
    if not found_student:
        print("Student Not Found")

while True:
    print()
    print("_________Student Management System Menu_________")
    print("SELECT AN OPTION YOU WANT")
    print("1.View Students")
    print("2. Search Student")
    print("3. Add Student")
    print("4. Update Grade")
    print("5. Delete Student")
    print("6. Exit")

    choice = input("Enter choice: ").strip()

    if choice == "1":
        view_students()

    elif choice == "2":
        search_student()

    elif choice == "3":
        add_student()

    elif choice == "4":
        update_grade()

    elif choice == "5":
        delete_student()

    elif choice == "6":
        print("Exit")
        break
    else:
        print("Invalid Choice")
    input("Press Enter to return to the menu...")
```
