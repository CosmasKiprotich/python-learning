```python
students_list = [
    {"name": "John", "grade": "A"},
    {"name": "Mary", "grade": "B"},
    {"name": "Peter", "grade": "C"},
    {"name": "Sarah", "grade": "A"},
    {"name": "David", "grade": "D"}
]
grade_counts = {
    "A": 0,
    "B": 0,
    "C": 0,
    "D": 0,
    "F": 0
}

for student in students_list:
    grade_counts[student["grade"]] += 1
print(grade_counts)

```

```python
def calculate_average(numbers):
    total = 0
    for number in numbers:
        total += number
    average = total / len(numbers)
    return average


numbers = [10, 20, 30, 40, 50]
result = calculate_average(numbers)
print(result)
```

```python
students = [
    {"id": "ST001", "name": "John"},
    {"id": "ST002", "name": "Mary"},
    {"id": "ST003", "name": "Peter"}
]


def find_student(students, student_id):
    for student in students:
        if student["id"] == student_id:
            return student

    return None


result = find_student(students, "ST002")
result1 = find_student(students, "ST009")

print(result)
print(result1)
```

```python
#linear search function

def linear_search(numbers, target):
    for number in numbers:
        if number == target:
            return True

    return False


numbers = [15, 8, 23, 42, 16, 4, 9]

print(linear_search(numbers, 42))
print(linear_search(numbers, 100))
```

```python
# Binary_search
# Binary_search requires data to be sorted

def binary_search(numbers, target):
    numbers.sort()
    left, right = 0, len(numbers) - 1
    while left <= right:
        mid = (left + right) // 2
        if numbers[mid] == target:
            return True
        elif numbers[mid] < target:
            left = mid + 1
        else:
            right = mid - 1
    return False
numbers = [15, 8, 23, 42, 16, 4, 9]
#numbers = [3, 7, 12, 18, 25, 31, 42, 50]

print(binary_search(numbers, 4))
print(binary_search(numbers, 20))
```

```python
#bubble_sort
#generally uses two nested loops
numbers = [5, 2, 8, 1, 3]

for i in range(len(numbers)):
    for j in range(0, len(numbers) - i -1):
        if numbers[j] > numbers[j + 1]:
            numbers[j], numbers[j + 1] = numbers[j + 1], numbers[j]

print(numbers)
```

```python
#buble_sort function

def bubble_sort(numbers):
    for i in range(len(numbers)):
        for j in range(0, len(numbers) - i -1):
            if numbers[j] > numbers[j + 1]:
                numbers[j], numbers[j + 1] = numbers[j + 1], numbers[j]
    return numbers
numbers = [15, 8, 23, 42, 16, 4, 9]

print(bubble_sort(numbers))
```

```python
#Selection_sort
numbers = [29, 10, 14, 37, 13]

def selection_sort(numbers):
    for i in range(len(numbers)):
        minimum_index = i
        for j in range(i + 1, len(numbers)):
            if numbers[j] < numbers[minimum_index]:
                minimum_index = j
        numbers[i], numbers[minimum_index] = numbers[minimum_index], numbers[i]
    return numbers

print(selection_sort(numbers))
```

```python
#continous selection sort

numbers = [64, 25, 12, 22, 11]

def selection_sort1(numbers):
    for i in range(len(numbers)):
        min_index = i
        for j in range(i + 1, len(numbers)):
            if numbers[j] < numbers[min_index]:
                min_index = j
        numbers[i], numbers[min_index] = numbers[min_index], numbers[i]

    return numbers
print(selection_sort1(numbers))
```

```python
#insertion sort

numbers = [5, 2, 4, 6, 1, 3]

def insertion_sort(numbers):
    for i in range(len(numbers)):
        j = i
        while j > 0 and numbers[j - 1] > numbers[j]:
            numbers[j], numbers[j - 1] = numbers[j - 1], numbers[j]
            j -= 1
    return numbers

print(insertion_sort(numbers))
```

```python
numbers = [5, 2, 4, 6, 1, 3]

def insertion_sort(numbers):
    for i in range(1, len(numbers)):
        key = numbers[i]
        j = i - 1
        while j >= 0 and numbers[j] > key:
            numbers[j + 1] = numbers[j]
            j -= 1
        numbers[j + 1] = key
    return numbers
print(insertion_sort(numbers))
```
